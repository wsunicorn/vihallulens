# T24 / E05 — Đối chứng cửa sổ không chồng lấn

Một lượt chạy, **~64 phút**, để tách một câu hỏi mà T23 không trả lời được.

## Câu hỏi

T23 cho thấy chia theo câu thắng cả ba cỡ cửa sổ. Nhưng ba cỡ ấy đều **chồng lấn** (bước bằng
nửa cửa sổ) còn chia theo câu thì **phủ kín và không đè lên nhau**. Nên "câu thắng" còn hai cách
giải thích, và T23 không phân biệt được:

1. **Ngữ nghĩa** — ranh giới câu mang thông tin mà ranh giới token tùy tiện không có.
2. **Chồng lấn** — một token trong vùng chồng bị đếm cho hai đoạn, làm véc-tơ theo đoạn không
   còn là một phân bố theo nghĩa chặt, và chính điều đó làm hỏng đặc trưng hình dạng.

Nếu là (2) thì phần đóng góp của đề tài phải viết lại: nó không phải "chia theo đơn vị nghĩa" mà
chỉ là "đừng chia chồng lấn".

## Cách tách

Khóa cả hai biến gây nhiễu, để hai cấu hình khác nhau **đúng một thứ**:

| | TB đoạn | trung vị | chồng lấn | ranh giới |
|---|---|---|---|---|
| Chia theo câu (E03) | 5,29 | 5 | không | **ngữ nghĩa** |
| **Cửa sổ 48, bước 48** | **5,56** | **5** | không | tùy tiện |

Cỡ 48 chọn bằng đo trên CPU chứ không đoán. Phương án nghĩ tới đầu tiên — cửa sổ 128 bước 128 —
chỉ cho **2,43** đoạn, tức lại lẫn số đoạn vào và là một đối chứng hỏng.

## Đọc kết quả thế nào — viết trước khi thấy số

Mốc so là **0,7768**, điểm dev của E03 chấm **trên cùng máy sẽ chấm lượt này**.

- **Câu vẫn hơn rõ** → kết luận là **ngữ nghĩa**. Đóng góp của đề tài đứng vững như đang viết.
- **Hòa, chênh dưới ~0,008** → không kết luận được. Biên độ trôi giữa hai môi trường đo ở T23 đã
  là 0,0075, và dev chỉ có 700 mẫu. Phải viết là "không phân biệt được", không được chọn cách
  đọc nào có lợi hơn.
- **Cửa sổ 48 hơn** → khoảng cách ở T23 là do **chồng lấn**, không phải ranh giới câu. Phải sửa
  lại phần đóng góp trong báo cáo, và T24 phải chọn cửa sổ 48 chứ không chọn chia theo câu.

Trường hợp thứ ba là trường hợp bất lợi nhất cho đề tài. Ghi ra đây trước khi chạy để sau không
đọc trại đi.

```python
# Ô 1 — lấy code. Chạy lại được nhiều lần.
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/wsunicorn/vihallulens.git"
REPO_DIR = Path("/kaggle/working/vihallulens")


def run(*args, cwd=None):
    done = subprocess.run(args, cwd=cwd, capture_output=True, text=True)
    if done.returncode:
        raise RuntimeError(" ".join(args) + chr(10) + done.stdout + done.stderr)
    return done.stdout.strip()


if (REPO_DIR / ".git").is_dir():
    run("git", "fetch", "--quiet", "origin", cwd=REPO_DIR)
    run("git", "reset", "--quiet", "--hard", "origin/main", cwd=REPO_DIR)
    print("đã cập nhật repo có sẵn")
else:
    run("git", "clone", "--quiet", REPO_URL, str(REPO_DIR))
    print("đã clone mới")

%cd /kaggle/working/vihallulens
print("commit:", run("git", "log", "--oneline", "-1", cwd=REPO_DIR))
```

```python
# Ô 4 — xác nhận cỡ 48 thật sự khớp mật độ đoạn của chia theo câu. Khoảng 1 phút, CPU.
# Đây là tiền đề của cả thí nghiệm: lệch mật độ thì đối chứng mất nghĩa và không nên chạy tiếp.
import sys

sys.path.insert(0, "src")
import numpy as np
import pandas as pd
from transformers import AutoTokenizer

from vihallulens.data.chunking import chunk_context

tok = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-7B-Instruct")
ctx = pd.concat(
    [pd.read_parquet(f"data/interim/vihallu_{s}.parquet") for s in ("train", "dev", "test")]
)["context"].tolist()

sent = np.array([len(chunk_context(c, strategy="sentence", min_words=5)) for c in ctx])
win = np.array([
    len(chunk_context(c, strategy="token_window", tokenizer=tok, window_size=48, stride=48))
    for c in ctx
])
print(f"chia theo câu     : TB {sent.mean():.2f}  trung vị {np.median(sent):.0f}")
print(f"cửa sổ 48 bước 48 : TB {win.mean():.2f}  trung vị {np.median(win):.0f}")
print(f"lệch mật độ       : {win.mean() - sent.mean():+.2f} đoạn")
assert abs(win.mean() - sent.mean()) < 0.5, "lệch mật độ quá lớn, đối chứng mất nghĩa"
print("\nĐạt: hai cách chia khớp mật độ, khác nhau đúng ở ranh giới ngữ nghĩa.")
```

```python
# Ô 8 — kiểm toàn vẹn shard trước khi rời phiên. Vài giây, CPU.
# Bài học T23: sau vài giờ GPU mà không có gì khẳng định shard đủ dòng và đọc được. Ô này chạy
# lúc còn kịp trích lại, thay vì phát hiện ra khi đã về nhà.
#
# Chỉ kiểm ĐÚNG hai shard của lượt này, tìm theo hash trích trong cấu hình. Quét cả thư mục sẽ
# báo động giả trên shard cũ của T20 — chúng chỉ có 2 khối vì lúc ấy chưa có đặc trưng chunk,
# và đó là bình thường chứ không phải hỏng.
import json
import sys
from pathlib import Path

sys.path.insert(0, "src")
from vihallulens.config import extraction_hash, load_config

run = extraction_hash(load_config("configs/e05_control_window48_vihallu.yaml"))
expected = {"train": 5600, "dev": 700}
ok = True
for split, want in expected.items():
    path = Path("data/processed") / f"vihallu_{split}_{run}.jsonl"
    if not path.exists():
        print(f"  THIẾU {path.name}")
        ok = False
        continue
    rows = [json.loads(line) for line in path.open(encoding="utf-8") if line.strip()]
    ids = {r["sample_id"] for r in rows}
    blocks = [k for k in rows[0] if k.startswith(("lookback_", "chunk_", "top1_"))]
    chunks = [r["n_chunks"] for r in rows]
    good = len(rows) == want and len(ids) == want and len(blocks) == 7
    ok &= good
    print(f"  {'OK ' if good else 'HỎNG'} {path.name:<44} {len(rows):>5}/{want} dòng, "
          f"{len(ids):>5} id, {len(blocks)} khối, TB {sum(chunks) / len(chunks):.2f} đoạn")
print("\nHai shard hợp lệ." if ok else "\nCÓ VẤN ĐỀ — chạy lại ô trích trước khi rời phiên.")
```
